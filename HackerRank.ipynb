```python
#Capitalize!
def solve(s):
    for word in s.split(' '):
        s = s.replace(word, word.capitalize())
    return s
print(solve("moaz ahmed mohamed elsaeed"))    
```

```python
#Lists
N = int(input())
lst = []

for _ in range(N):
    command = input().split()
    action = command[0]

    if action == "insert":
        lst.insert(int(command[1]), int(command[2]))
    elif action == "print":
        print(lst)
    elif action == "remove":
        lst.remove(int(command[1]))
    elif action == "append":
        lst.append(int(command[1]))
    elif action == "sort":
        lst.sort()
    elif action == "pop":
        lst.pop()
    elif action == "reverse":
        lst.reverse()       
```

```python
#Find a string
def count_substring(string, sub_string):
    count = 0
    for i in range(len(string)):
        if string[i:i+len(sub_string)] == sub_string:
            count += 1
    return count

string = input()
sub_string = input()

print(count_substring(string, sub_string))
```

```python
#sWAP cASE
def swap_case(s):
    new_s = ""
    for x in s:
        if x.isupper():
            new_s += x.lower()
        else:
            new_s += x.upper()
    return new_s

s = input()
print(swap_case(s))
```

```python
#Detect Floating Point Number
import re

def is_valid_float(s):
    pattern = r'^[+-]?[0-9]*\.[0-9]+$'
    return bool(re.match(pattern, s))

t = int(input())

for _ in range(t):
    s = input().strip()
    print(is_valid_float(s))
```

```python
#Merge the Tools!
def merge_the_tools(string, k):
    for i in range(0, len(string), k):
        sub = string[i:i + k]

        seen = set()
        result = []

        for char in sub:
            if char not in seen:
                seen.add(char)
                result.append(char)

        print("".join(result))
string = input()
k = int(input())
merge_the_tools(string, k)
```

```python
#Introduction to Sets
def average(array):
    distinct_heights = set(array)
    return round(sum(distinct_heights) / len(distinct_heights), 3)

n = int(input())
arr = list(map(int, input().split()))

print(average(arr))
```

```python
#Time Delta
from datetime import datetime

def time_delta(t1, t2):
    fmt = "%a %d %b %Y %H:%M:%S %z"

    dt1 = datetime.strptime(t1, fmt)
    dt2 = datetime.strptime(t2, fmt)

    diff_seconds = int(abs((dt1 - dt2).total_seconds()))

    return str(diff_seconds)


t1 = input()
t2 = input()

print(time_delta(t1, t2))
```

```python
#Iterables and Iterators
from itertools import combinations

n = int(input())
letters = input().split()
k = int(input())

all_combs = list(combinations(letters, k))
count_a = sum(1 for comb in all_combs if 'a' in comb)

print(count_a / len(all_combs))
```

```python
#Maximize It!
from itertools import product

k, m = map(int, input().split())

lists = []
for _ in range(k):
    lists.append(list(map(int, input().split()))[1:])

max_val = 0

for comb in product(*lists):
    value = sum(x**2 for x in comb) % m
    max_val = max(max_val, value)

print(max_val)
```
